```python
import heapq


def a_star_search(graph, start, goal, heuristic):
    """
    Performs A* Search on a graph.

    graph: Dictionary where keys are nodes,
           and values are lists of (neighbor, edge_cost) tuples.

    start: Starting node
    goal: Destination node
    heuristic: Estimated cost from each node to the goal
    """

    # Priority queue stores:
    # (f_cost, g_cost, current_node, path)

    # Initially:
    # g(start) = 0
    # f(start) = g(start) + h(start)

    pqueue = [(heuristic[start], 0, start, [start])]

    # Store the minimum cost found to reach each node
    visited = {}

    while pqueue:

        # Remove node with lowest f(n)
        f_cost, cost, current, path = heapq.heappop(pqueue)

        # Goal test
        if current == goal:
            return path, cost

        # Skip if we already found a cheaper path
        if current in visited and visited[current] <= cost:
            continue

        visited[current] = cost

        # Explore neighbours
        for neighbor, edge_cost in graph.get(current, []):

            # g(n) = actual cost from start to neighbour
            new_cost = cost + edge_cost

            # h(n) = heuristic cost from neighbour to goal
            h_cost = heuristic[neighbor]

            # f(n) = g(n) + h(n)
            new_f_cost = new_cost + h_cost

            # Add neighbour to priority queue
            if neighbor not in visited or new_cost < visited[neighbor]:

                heapq.heappush(
                    pqueue,
                    (
                        new_f_cost,
                        new_cost,
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, float('inf')


# -------------------------------------------------
# Example Usage
# -------------------------------------------------

if __name__ == "__main__":

    # Graph:
    # Node -> [(Neighbour, Edge Cost)]

    graph = {
        'A': [('B', 2), ('C', 5)],
        'B': [('D', 4), ('E', 2)],
        'C': [('E', 1), ('F', 8)],
        'D': [('G', 3)],
        'E': [('G', 6), ('D', 1)],
        'F': [('G', 3)],
        'G': []
    }

    # Heuristic distance from each node to goal G
    heuristic = {
        'A': 7,
        'B': 5,
        'C': 6,
        'D': 3,
        'E': 4,
        'F': 3,
        'G': 0
    }

    start_node = 'A'
    goal_node = 'G'

    path, total_cost = a_star_search(
        graph,
        start_node,
        goal_node,
        heuristic
    )

    if path:
        print(f"Optimal Path: {' -> '.join(path)}")
        print(f"Total Cost: {total_cost}")
    else:
        print("No path found.")
```
