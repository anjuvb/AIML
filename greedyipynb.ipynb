```python

import heapq

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

h = {
    'A': 6,
    'B': 4,
    'C': 2,
    'D': 5,
    'E': 3,
    'F': 1,
    'G': 0
}

def greedy_best_first_search(start, goal):
    pq = [(h[start], start)]
    visited = set()
    parent = {start: None}

    while pq:
        _, node = heapq.heappop(pq)

        if node in visited:
            continue

        visited.add(node)
        print("Visiting:", node)

        if node == goal:
            path = []
            while node is not None:
                path.append(node)
                node = parent[node]
            print("Path:", path[::-1])
            return

        for neighbor in graph[node]:
            if neighbor not in visited:
                parent[neighbor] = node
                heapq.heappush(pq, (h[neighbor], neighbor))

    print("Goal not found")

greedy_best_first_search('A', 'G')
```
