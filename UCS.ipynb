```python
import heapq

def uniform_cost_search(graph, start, goal):
    # Priority queue: (cost, current_node, path)
    queue = [(0, start, [start])]
    visited = set()

    while queue:
        cost, node, path = heapq.heappop(queue)

        if node in visited:
            continue

        visited.add(node)

        # Goal reached
        if node == goal:
            return cost, path

        # Explore neighbours
        for neighbour, edge_cost in graph[node]:
            if neighbour not in visited:
                heapq.heappush(
                    queue,
                    (cost + edge_cost, neighbour, path + [neighbour])
                )

    return None, None


# Graph
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('C', 2), ('D', 5)],
    'C': [('D', 1)],
    'D': [('E', 3)],
    'E': []
}

start = 'A'
goal = 'E'

cost, path = uniform_cost_search(graph, start, goal)

print("Shortest path:", path)
print("Total cost:", cost)
```
