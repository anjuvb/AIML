4) Detect the missing values

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

#Fill missing values in numerical column with their median
for col in num_cols:
  df[col] = df[col].fillna(df[col].median())

#Fill missing values in categorical columns with their mode
for col in cat_cols:
  df[col] = df[col].fillna(df[col].mode()[0])

#Verify whether any missing values remain
remaining = df.isnull().sum()
remaining[remaining > 0]
```

5)Identified and removed duplicate records.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Convert numerical columns to numeric (if needed)
numeric_columns = [
    "popularity",
    "acousticness",
    "danceability",
    "duration_ms",
    "energy",
    "instrumentalness",
    "liveness",
    "loudness",
    "speechiness",
    "tempo",
    "valence"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

# Draw Box Plots
for col in numeric_columns:
    plt.figure(figsize=(5,4))
    plt.boxplot(df[col].dropna())
    plt.title(f"Box Plot of {col}")
    plt.ylabel(col)
    plt.grid(True)
    plt.show()

# Count Outliers using IQR
for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

6) Detect outliers using statistical methods (IQR or Z-score) and visualize them
using box plots.

```python
import matplotlib.pyplot as plt

# Numerical columns
columns = [
    "popularity",
    "acousticness",
    "danceability",
    "duration_ms",
    "energy",
    "instrumentalness",
    "liveness",
    "loudness",
    "speechiness",
    "tempo",
    "valence"
]

# Handle outliers using IQR (Capping)
for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outliers handled successfully.")
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Columns to scale
columns = [
    "popularity",
    "acousticness",
    "danceability",
    "duration_ms",
    "energy",
    "instrumentalness",
    "liveness",
    "loudness",
    "speechiness",
    "tempo",
    "valence"
]

# ---------------- Min-Max Normalization ----------------
minmax = MinMaxScaler()

df_minmax = df.copy()

df_minmax[[col + "_MinMax" for col in columns]] = minmax.fit_transform(df[columns])

display(df_minmax[[columns[0], columns[0] + "_MinMax",
                   columns[1], columns[1] + "_MinMax"]].head())

print("\n")

# ---------------- Standardization ----------------
standard = StandardScaler()

df_standard = df.copy()

df_standard[[col + "_Standard" for col in columns]] = standard.fit_transform(df[columns])

display(df_standard[[columns[0], columns[0] + "_Standard",
                     columns[1], columns[1] + "_Standard"]].head())
```
